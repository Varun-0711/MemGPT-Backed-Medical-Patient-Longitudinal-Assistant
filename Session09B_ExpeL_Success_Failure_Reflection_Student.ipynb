### Activity 1 (Tier 1 - Warmup): The Trajectory Data Structure
**Concept**: An episodic trial logs the goal, actions, error message (if any), and success boolean.
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
@dataclass
class Trajectory:
    task: str
    actions: list[str]
    success: bool
    error: str = ""

t_failed = Trajectory(
    task="Query monthly active users in SQL",
    actions=["SELECT COUNT(user_id) FROM logins WHERE date >= '2024-01-01'"],
    success=False,
    error="Column user_id contains duplicate logins; returns inflated count"
)
t_success = Trajectory(
    task="Query monthly active users in SQL",
    actions=["SELECT COUNT(DISTINCT user_id) FROM logins WHERE date >= '2024-01-01'"],
    success=True
)
print("Failed Trial:", t_failed)
print("Success Trial:", t_success)
```

### Activity 2 (Tier 1 - Warmup): Generative Agents Memory Scoring
**Concept**: Tripartite ranking score:
$$Score = lpha \cdot Recency + eta \cdot Importance + \gamma \cdot Relevance$$
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
def score_memory(recency_hours: float, importance_1_to_10: float, sim: float,
                 alpha=1.0, beta=1.0, gamma=1.0) -> float:
    # Recency decay (0.995^hours)
    recency_decay = 0.995 ** recency_hours
    norm_importance = importance_1_to_10 / 10.0
    return alpha * recency_decay + beta * norm_importance + gamma * sim

print("Score for memory (1 hour ago, imp=9, sim=0.85):",
      score_memory(recency_hours=1, importance_1_to_10=9, sim=0.85))
```
