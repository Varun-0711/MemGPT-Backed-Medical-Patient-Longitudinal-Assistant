### Activity 1 (Tier 1 - Warmup): Entity-Relation-Entity Triple Extraction
**Concept**: Knowledge graphs represent information as directed labeled edges: `(Subject, Predicate, Object)`.
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
sample_text = "Dr. Anant leads the Agentic AI curriculum. The curriculum uses MemGPT and Voyager."

# Minimal rule-based triple extractor for demonstration
def extract_triples(text: str) -> list[tuple[str, str, str]]:
    triples = [
        ("Dr. Anant", "LEADS", "Agentic AI Curriculum"),
        ("Agentic AI Curriculum", "USES", "MemGPT"),
        ("Agentic AI Curriculum", "USES", "Voyager")
    ]
    return triples

triples = extract_triples(sample_text)
print("Extracted Knowledge Triples:")
for s, p, o in triples:
    print(f"  ({s}) ---[{p}]---> ({o})")
```

---
### Activity 2 (Tier 1 - Warmup): Graph Adjacency List & Node Degree
**Concept**: A graph is stored as an adjacency list mapping nodes to their outgoing edges.
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
class SimpleKnowledgeGraph:
    def __init__(self):
        self.adj = defaultdict(list)
        self.nodes = set()

    def add_edge(self, src: str, rel: str, dst: str):
        self.nodes.add(src)
        self.nodes.add(dst)
        self.adj[src].append((rel, dst))

    def degree(self, node: str) -> int:
        return len(self.adj[node])

kg = SimpleKnowledgeGraph()
for s, p, o in triples:
    kg.add_edge(s, p, o)

print("Nodes in Knowledge Graph:", kg.nodes)
print("Degree of 'Agentic AI Curriculum':", kg.degree("Agentic AI Curriculum"))
```
