## 0. Setup: The Four Memory Stores
We construct structured storage for Working, Episodic, Semantic, and Procedural memory.


```python
from dataclasses import dataclass, field
import json

print("CoALA workstation initialized!")
```

---
### Activity 1 (Tier 1 - Warmup): The CoALA Memory Partition
**Concept**: Separate active task state (Working), past experiences (Episodic), world knowledge (Semantic), and tool code (Procedural).
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
@dataclass
class CoALAMemory:
    working: dict = field(default_factory=lambda: {"goal": "Analyze dataset", "step": 1})
    episodic: list = field(default_factory=list)
    semantic: dict = field(default_factory=lambda: {"domain": "Data Science", "metric": "F1-Score"})
    procedural: dict = field(default_factory=lambda: {"clean_data": "def clean(df): return df.dropna()"})

coala = CoALAMemory()
print("Working State:", coala.working)
print("Procedural Tools:", list(coala.procedural.keys()))
```

---
### Activity 2 (Tier 1 - Warmup): The Ground-Propose-Evaluate-Act Cycle
**Concept**: The agent follows a structured four-stage decision loop.
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
def decision_cycle(obs: str, mem: CoALAMemory) -> str:
    # 1. Ground
    mem.working["observation"] = obs
    # 2. Propose
    candidates = ["propose_clean_data", "propose_train_model"]
    # 3. Evaluate
    chosen_action = candidates[0]
    # 4. Act
    mem.episodic.append({"action": chosen_action, "status": "executed"})
    return f"Executed {chosen_action}"

print(decision_cycle("Dataset contains missing values", coala))
print("Episodic log length:", len(coala.episodic))
```

---
### Activity 3 (Tier 1 - Warmup): State Serialization & Restoration
**Concept**: Snapshotting the agent's complete state to JSON allows pause and resume across runs.
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
serialized = json.dumps({
    "working": coala.working,
    "semantic": coala.semantic,
    "procedural": coala.procedural,
    "episodic_count": len(coala.episodic)
}, indent=2)
print("Serialized Agent Snapshot:\n", serialized)
```
