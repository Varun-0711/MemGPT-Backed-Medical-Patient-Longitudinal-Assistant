### Activity 2 (Tier 1 - Warmup): Conformal Prediction Set Construction
**Concept**: A conformal prediction set includes all actions with confidence $\ge 	au$. If set size $> 1$, the situation is ambiguous!
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
def build_conformal_set(actions_prob: dict[str, float], threshold: float = 0.20) -> list[str]:
    return [act for act, p in actions_prob.items() if p >= threshold]

conf_set = build_conformal_set(candidate_actions, threshold=0.20)
print("Conformal Action Set:", conf_set)
print(f"Set size: {len(conf_set)} -> {'AMBIGUOUS (Needs Clarification)' if len(conf_set) > 1 else 'UNAMBIGUOUS'}")
```
