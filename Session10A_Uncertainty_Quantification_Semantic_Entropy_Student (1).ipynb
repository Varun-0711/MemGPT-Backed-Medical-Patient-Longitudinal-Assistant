### Activity 2 (Tier 1 - Warmup): Token-Level Shannon Entropy
**Concept**: Shannon entropy over token probabilities:
$$H(X) = - \sum_{i} P(x_i) \log_2 P(x_i)$$
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
def shannon_entropy(probs: list[float]) -> float:
    return -sum(p * math.log2(p) for p in probs if p > 0)

# Confident distribution (90% / 10%) vs Uncertain (50% / 50%)
print("Confident Entropy (0.9, 0.1):", shannon_entropy([0.9, 0.1]))
print("Uniform Uncertain Entropy (0.5, 0.5):", shannon_entropy([0.5, 0.5]))
```
