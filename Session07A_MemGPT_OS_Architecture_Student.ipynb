### Activity 1 (Tier 1 - Warmup): The CoreMemory Dataclass
**Concept**: Core Memory holds mutable `persona` and `human` scratchpad blocks pinned inside the system prompt.
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
@dataclass
class CoreMemory:
    persona: str = "You are an autonomous research assistant. You answer factually."
    human: str = "User Name: Saarang\nRole: AI Engineer\nLocation: Coimbatore"

    def compile_prompt(self) -> str:
        return f"=== CORE MEMORY ===\n[PERSONA]\n{self.persona}\n\n[HUMAN]\n{self.human}\n==================="

mem = CoreMemory()
print(mem.compile_prompt())
```

---
### Activity 2 (Tier 1 - Warmup): Memory Mutation Tool (core_memory_append)
**Concept**: Agents mutate their own core memory blocks using tool calls.
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
def tool_core_memory_append(core_mem: CoreMemory, block: str, fact: str) -> str:
    if block == "human":
        core_mem.human += f"\n• {fact.strip()}"
        return f"SUCCESS: Appended fact to human block."
    elif block == "persona":
        core_mem.persona += f"\n• {fact.strip()}"
        return f"SUCCESS: Appended fact to persona block."
    return "ERROR: Invalid block name. Use 'human' or 'persona'."

status = tool_core_memory_append(mem, "human", "Preferred language: Python 3.12")
print(status)
print("\nUpdated Core Memory:\n", mem.compile_prompt())
```

### Activity 3 (Tier 1 - Warmup): Context Token Counter & Budget Watchdog
**Concept**: The agent must track its active token budget and trigger warnings before overflow.
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
def check_context_budget(context_text: str, budget_limit: int = 500) -> dict:
    # Approximate token count (words * 1.3)
    est_tokens = int(len(context_text.split()) * 1.3)
    usage_pct = (est_tokens / budget_limit) * 100.0
    return {
        "estimated_tokens": est_tokens,
        "budget_limit": budget_limit,
        "usage_pct": round(usage_pct, 1),
        "status": "CRITICAL" if usage_pct > 85 else ("WARNING" if usage_pct > 70 else "NORMAL")
    }

print("Budget check:", check_context_budget(mem.compile_prompt()))
```
