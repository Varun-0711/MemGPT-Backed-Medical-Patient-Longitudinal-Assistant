---
### Activity 2 (Tier 1 - Warmup): Precision@1 and Exact Match Scoring
**Concept**: Evaluating whether an agent's recalled answer contains the planted ground truth.
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
def score_recall(predicted: str, gold: str) -> bool:
    return gold.lower() in predicted.lower()

predictions = [
    "You are an AI Engineer working in tech.",
    "You are based in Coimbatore.",
    "You enjoy writing in Python 3.12."
]

results = [score_recall(p, f["gold"]) for p, f in zip(predictions, planted_facts)]
print("Recall Accuracy:", sum(results) / len(results) * 100, "%")
```

### Activity 3 (Tier 1 - Warmup): Token Expenditure Accounting
**Concept**: Tracking token costs across turns to measure economic efficiency.
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
class TokenTracker:
    def __init__(self, cost_per_1k: float = 0.0005):
        self.total_tokens = 0
        self.cost_per_1k = cost_per_1k

    def log(self, tokens: int):
        self.total_tokens += tokens

    def total_cost(self) -> float:
        return (self.total_tokens / 1000.0) * self.cost_per_1k

tracker = TokenTracker()
tracker.log(1200)
tracker.log(850)
print(f"Total Tokens: {tracker.total_tokens} | Estimated Cost: ${tracker.total_cost():.6f}")
```
