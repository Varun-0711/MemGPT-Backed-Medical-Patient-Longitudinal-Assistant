## 0. Setup: The Sandboxed Skill Execution Engine
We build an executable skill library storing Python functions as verified source strings.


```python
import math
from collections import defaultdict

print("Voyager Skill Library Workstation initialized!")
```

---
### Activity 1 (Tier 1 - Warmup): The Executable Skill Library Store
**Concept**: A skill is stored as executable source code along with its docstring description and signature.
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
class SkillLibrary:
    def __init__(self):
        self.skills = {}

    def add_skill(self, name: str, docstring: str, code: str):
        self.skills[name] = {"doc": docstring, "code": code}

    def retrieve(self, query: str) -> dict:
        # Simple keyword overlap matcher for demonstration
        best_name = None
        best_score = -1
        for name, data in self.skills.items():
            score = sum(1 for w in query.lower().split() if w in data["doc"].lower())
            if score > best_score:
                best_score = score
                best_name = name
        return {best_name: self.skills[best_name]} if best_name else {}

library = SkillLibrary()
library.add_skill("calc_compound_interest", "Calculates compound interest given principal, rate, and time.",
                  "def calc_compound_interest(p, r, t): return p * (1 + r)**t")
print("Retrieved skill for 'compound interest':", library.retrieve("compound interest"))
```

---
### Activity 2 (Tier 1 - Warmup): Sandboxed Execution of Stored Skills
**Concept**: Executing retrieved Python code dynamically inside an isolated dictionary namespace.
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
def execute_skill(code_str: str, fn_name: str, *args):
    namespace = {}
    exec(code_str, namespace)
    target_fn = namespace[fn_name]
    return target_fn(*args)

code = library.skills["calc_compound_interest"]["code"]
result = execute_skill(code, "calc_compound_interest", 1000, 0.05, 3)
print(f"Executed skill result: Principal $1000 at 5% for 3 years = ${result:.2f}")
```

---
### Activity 3 (Tier 1 - Warmup): Skill Composition: Chaining Primitives
**Concept**: Combining two acquired skills to solve a composite task.
*Note: This foundational activity is pre-solved for you below.*


```python
# Pre-solved warmup
library.add_skill("clean_currency", "Strips dollar sign and commas from currency string.",
                  "def clean_currency(s): return float(s.replace('$', '').replace(',', ''))")

# Composite execution: Clean string -> Compute interest
raw_principal = "$10,000"
p_clean = execute_skill(library.skills["clean_currency"]["code"], "clean_currency", raw_principal)
final_val = execute_skill(library.skills["calc_compound_interest"]["code"], "calc_compound_interest", p_clean, 0.08, 5)
print(f"Composite Pipeline: {raw_principal} -> Clean: {p_clean} -> Final 5yr: ${final_val:.2f}")
```
